###Read the data from silver layer

##Firstway

```python
df = spark.table('etl_project.silver.silver_cust_info')
df.display()
```

###Another way

```python
query = """
    SELECT
        customer_key,
        customer_id,
        first_name,
        last_name,
        marital_status
    FROM etl_project.silver.silver_cust_info
"""

df = spark.sql(query)
```

```python
df.display()
```

#write to the gold layer

```python
df.write.mode("overwrite").format("delta").saveAsTable("etl_project.gold.gold_cust_info")
```
